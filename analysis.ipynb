```python
# importing libraries
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
# Loading Data
df = pd.read_csv("employees.csv")
df
```

```python
# Average salary
df["salary"].mean()
```

```python
# Average salary per department
avg_salary = df.groupby("dept_id", as_index=False)["salary"].mean()
avg_salary
```

```python
# Employees per department
emp_count = df.groupby("dept_id")["id"].count()
emp_count
```

```python
# Visualization — Average Salary
avg_salary.plot(kind="bar", x="dept_id", y="salary")
plt.title("Average Salary by Department")
plt.xlabel("Department ID")
plt.ylabel("Average Salary")
plt.show()
```

```python
# Visualization — Employee Count
emp_count.plot(kind="bar")
plt.title("Number of Employees per Department")
plt.xlabel("Department ID")
plt.ylabel("Number of Employees")
plt.show()
```

Insights:
- Department 2 has the highest average salary.
- Department 1 has fewer employees compared to others.
- Salary distribution indicates higher cost concentration in Department 2.
